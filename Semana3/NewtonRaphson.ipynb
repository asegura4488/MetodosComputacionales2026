```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def Function(x):
  #return 5*(1-np.exp(-x)) - x
  return x**2 + 2
```

```python
def Derivative(x,f,h=1e-6):
  return (f(x+h)-f(x-h))/(2*h)
```

```python
def GetNewtonMethod(f,df,xn,itmax=1000,precision=1e-10):

  error = 1
  it = 0

  while error > precision and it < itmax:

    try:
      xn1 = xn - f(xn)/df(xn,f)
      error = np.abs( xn1 - xn )
    except ZeroDivisionError:
      print('Division por zero')

    xn = xn1
    it += 1

  if it == itmax:
    return False
  else:
    #print(it)
    return xn
```

```python
GetNewtonMethod(Function,Derivative,10.) # la semilla del algoritmo
```

```python
xtrial = np.linspace(-10,10,10)
for i in xtrial:
  print(GetNewtonMethod(Function,Derivative,i))
```
