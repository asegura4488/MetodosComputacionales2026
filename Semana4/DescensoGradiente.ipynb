```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time
import copy
```

```python
def Plotter(e):

  fig = plt.figure(figsize=(10,5))
  ax = fig.add_subplot(121)
  ax.set_title('epoch: {:.0f}'.format(e), fontsize=10)

  return fig, ax
```

```python
# Definimos una funcion no convexa
def f(x):
  #return x**2 + 10*np.sin(x)
  return np.sin(x)/x
```

```python
xvalues = np.linspace(-10,10,100)
yvalues = f(xvalues)
```

```python
# Definimos la derivada
def Derivative(x,f,h=1e-4):
  return (f(x+h) - f(x-h))/(2*h)
```

```python
def GradientDescent(f, epochs=200, lr=0.08):

  # Necesitamos una semilla para el metodos

  xn = np.random.uniform(-5,5)

  for e in range(epochs):
    clear_output(wait=True)

    # El algoritmo
    xn1 = xn - lr*Derivative(xn,f)
    # No se esta actualizando
    xn = xn1

    fig, ax = Plotter(e)
    ax.plot(xvalues,yvalues)
    ax.scatter(xn,f(xn),color='r')

    time.sleep(0.01)

    plt.show()
```

```python
GradientDescent(f)
```

```python
x = np.linspace(-2,2,100)
y = np.linspace(-2,2,100)
X,Y = np.meshgrid(x,y)
```

```python
def f2(x,y):
  return (x**2 - 1)**2 + y**2
```

```python
Mapa = np.zeros_like(X)
for i in range(len(x)):
  for j in range(len(y)):
    Mapa[i,j] = f2(x[i],y[j])
```

```python
fig = plt.figure()
ax = fig.add_subplot(211,projection='3d')
ax.plot_surface(X,Y,Mapa)
```
