```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sym
```

```python
X = np.array([1,2,3,4])
Y = np.array([0,2.5,1,4])
```

```python
fig = plt.figure(figsize=(4,4))
plt.scatter(X,Y,color='red')
```

```python
def Lagrange( x, X, i ):

  L = 1
  for j in range(len(X)):
    if i != j:
      L *= (x-X[j])/(X[i]-X[j])
  return L
```

```python
# Definir un simbolo
x= sym.Symbol('x',real=True)
x
```

```python
L0 = Lagrange(x,X,0)#.expand()
L0
```

```python
L1 = Lagrange(x,X,1)#.expand()
L1
```
